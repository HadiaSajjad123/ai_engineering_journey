```python
from pathlib import Path
import pandas as pd

data_path = Path("../data/credit_default.xls")

df = pd.read_excel(data_path, header=1)

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Work on a copy; keep the original df unchanged
df_clean = df.copy()

target = "default payment next month"

# Remove the customer identifier from model features later,
# but keep it in this working dataset for traceability.
print("Original shape:", df_clean.shape)
print("Missing values:", df_clean.isna().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
print("Duplicate IDs:", df_clean["ID"].duplicated().sum())

# Validate the target
print("Target values:", sorted(df_clean[target].unique()))
assert set(df_clean[target].unique()).issubset({0, 1})
assert df_clean["ID"].is_unique
assert df_clean[target].notna().all()
```

Handle Category Codes

```python
# Work on copies of the original category values
df_clean["EDUCATION"] = df_clean["EDUCATION"].replace({
    0: 4,
    5: 4,
    6: 4
})

df_clean["MARRIAGE"] = df_clean["MARRIAGE"].replace({
    0: 3
})

print("EDUCATION after recoding:")
display(df_clean["EDUCATION"].value_counts().sort_index())

print("MARRIAGE after recoding:")
display(df_clean["MARRIAGE"].value_counts().sort_index())
```

Engineering Financial Features

```python
# 1. Count months with a recorded payment delay
pay_cols = ["PAY_0", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6"]

df_clean["delayed_months"] = (df_clean[pay_cols] > 0).sum(axis=1)

# 2. Maximum recorded positive delay
df_clean["max_payment_delay"] = df_clean[pay_cols].clip(lower=0).max(axis=1)

# 3. Average bill amount over six months
bill_cols = [
    "BILL_AMT1", "BILL_AMT2", "BILL_AMT3",
    "BILL_AMT4", "BILL_AMT5", "BILL_AMT6"
]
df_clean["avg_bill_amount"] = df_clean[bill_cols].mean(axis=1)

# 4. Average payment amount over six months
payment_cols = [
    "PAY_AMT1", "PAY_AMT2", "PAY_AMT3",
    "PAY_AMT4", "PAY_AMT5", "PAY_AMT6"
]
df_clean["avg_payment_amount"] = df_clean[payment_cols].mean(axis=1)

# 5. Total payments over six months
df_clean["total_payment_amount"] = df_clean[payment_cols].sum(axis=1)

# 6. Recent bill relative to the granted credit limit
df_clean["recent_bill_to_limit_ratio"] = (
    df_clean["BILL_AMT1"] / df_clean["LIMIT_BAL"].replace(0, np.nan)
)

# 7. Recent payment relative to the absolute bill amount
df_clean["recent_payment_to_bill_ratio"] = (
    df_clean["PAY_AMT1"] / (df_clean["BILL_AMT1"].abs() + 1)
)

# Inspect the engineered features
new_features = [
    "delayed_months",
    "max_payment_delay",
    "avg_bill_amount",
    "avg_payment_amount",
    "total_payment_amount",
    "recent_bill_to_limit_ratio",
    "recent_payment_to_bill_ratio"
]

display(df_clean[new_features].describe().T)
```

Check New Features

```python
# Check engineered-feature quality
feature_check = pd.DataFrame({
    "Missing": df_clean[new_features].isna().sum(),
    "Unique Values": df_clean[new_features].nunique(),
    "Minimum": df_clean[new_features].min(),
    "Maximum": df_clean[new_features].max()
})

display(feature_check)

# Compare feature averages by target class
print("Feature averages by default outcome:")
display(
    df_clean.groupby(target)[new_features]
    .mean()
    .round(3)
    .T
)

# Default rate by number of delayed months
print("Default rate by delayed months:")
display(
    df_clean.groupby("delayed_months")[target]
    .agg(["count", "mean"])
    .rename(columns={"mean": "default_rate"})
    .assign(default_rate=lambda x: (x["default_rate"] * 100).round(2))
)
```

Save Processed Dataset Summary

```python
# Adjust this path if your notebook has a different working directory
data_dir = Path("week-02-loan-default-risk/data")
reports_dir = Path("week-02-loan-default-risk/reports")

data_dir.mkdir(parents=True, exist_ok=True)
reports_dir.mkdir(parents=True, exist_ok=True)

# Save processed data
processed_path = data_dir / "processed_credit_default.csv"
df_clean.to_csv(processed_path, index=False)

# Save a compact feature summary
summary_path = reports_dir / "day01_feature_summary.csv"

df_clean[new_features].describe().T.to_csv(summary_path)

print("Processed dataset saved to:", processed_path.resolve())
print("Feature summary saved to:", summary_path.resolve())
print("Processed dataset shape:", df_clean.shape)
```
